## Transformações geométricas
As transformações geométricas são aquelas cujo efeito é a **alteração da posição espacial dos pixels** que compõem uma imagem.
As principais operações são:
* Translação;
* Reflexão;
* Cisalhamento;
* Rotação;
* Escala.

```python
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np

mario = np.array(Image.open('../mario.png'))[:, :, :3]
(l, c, p) = mario.shape
print(mario.shape)
plt.imshow(mario)
```

## Translação
* Transladar uma imagem significa mover todos os pixels para uma nova coordenada;
* Pode ser realizada adicionando um valor às suas coordenadas.

$$\begin{pmatrix}
  x' \\ y'
\end{pmatrix} = \begin{pmatrix}
  x \\ y
\end{pmatrix} + \begin{pmatrix}
  d_x \\ d_y
\end{pmatrix}$$

### Translação (subarray, iterativo)

```python
(dx, dy) = (50, 40) # Deslocamento para a direita e para baixo.
(mario_start_x, mario_start_y, mario_end_x, mario_end_y) = (76, 60, 180, 180) # coordenadas do retângulo a ser copiado.
mario_tr = np.zeros(shape=(l + dy, c + dx, p), dtype=np.uint8) # Fundo preto por conta da deslocação.

# iteração sobre os pixels do retângulo a ser copiado, e cópia para a nova posição.
for i in range(mario_start_y, mario_end_y): 
    for j in range(mario_start_x, mario_end_x): 
        new_x = j + dx # coordenada x do pixel na nova posição.
        new_y = i + dy # coordenada y do pixel na nova posição.
        mario_tr[new_y, new_x ] = mario[i, j] # cópia do pixel para a nova posição de fato.

plt.figure(figsize=(8, 8)) # figura com 8 polegadas de largura e altura.
plt.subplot(2, 2, 1) #(nrows,ncol,index)
plt.imshow(mario) # plot da imagem orignal
plt.subplot(2, 2, 2) # (nrows,ncol,index) para a imagen deslocacada
plt.imshow(mario_tr) # plot da imagem transformada
```

### Translação (subarray, NumPy)

```python
(dx, dy) = (50, 40)
(mario_start_x, mario_start_y, mario_end_x, mario_end_y) = (76, 60, 180, 180) # coordenadas original do retângulo a ser copiado.
mario_tr = np.zeros(shape=(l + dy, c + dx, p), dtype=np.uint8) # Fundo preto.
mario_tr[mario_start_y + dy:mario_end_y + dy,
            mario_start_x + dx:mario_end_x + dx] = mario[mario_start_y:mario_end_y,
                                                         mario_start_x:mario_end_x]
plt.imshow(mario_tr)
```

### Translação (imagem inteira, iterativo)

```python
(dx, dy) = (50, 40) # tamanho do deslocamento
mario_tr = np.zeros(shape=(l + dy, c + dx, p), dtype=np.uint8) # onde l, c, p são as dimensões da imagem original. 
# Iteração sobre os pixels da imagem original, e cópia para a nova posição.
for i in range(l):
    for j in range(c):
        new_x = j + dx
        new_y = i + dy
        mario_tr[new_y, new_x] = mario[i, j]

plt.imshow(mario_tr)
```

### Translação (imagem inteira, NumPy)

```python
(dx, dy) = (50, 40)
mario_tr = np.zeros(shape=(l + dy, c + dx, p), dtype=np.uint8)
mario_tr[dy:l + dy, dx:c + dx] = mario # ou mario[:, :]
plt.imshow(mario_tr)
```

## Reflexão
* Refletir uma imagem é espelhar ela em relação a um eixo;
* Em outras palavras, a imagem resultante é como se a original fosse vista reproduzida por um espelho.

```python
# Reflexão em y:
# cria imagem vazia com o tamanho da imagem original
# a imagem recebe os pixels invertidos
mario_refl = np.zeros(shape=mario.shape, dtype=np.uint8)
for i in range(l):
    for j in range(c):
        new_x = -j # reflexão em y, então a coordenada x é invertida.
        new_y = i # a coordenada y permanece a mesma.
        mario_refl[new_y, new_x] = mario[i, j]

plt.imshow(mario_refl)
```

```python
# Reflexão em x:
mario_refl = np.zeros(shape=mario.shape, dtype=np.uint8)
# Acesso a índices: começo:fim:passo
# Logo: ::-1 significa do começo ao fim com passo -1 (de trás para frente)
mario_refl[::-1, ::1] = mario # ou mario[:, :] onde :: significa do começo ao fim e -1 significa do fim para o começo
plt.imshow(mario_refl)
```

## Cisalhamento
* Cisalhar uma imagem é realizar uma distorção nela, dando um efeito de *itálico*;
* A distorção ocorre ponto-a-ponto, ao aplicar-se um deslocamento à coordenada x ou y proporcional ao valor da coordenada y ou x, respectivamente;
* Isto é, y' é influenciado por x e x' é influenciado por y.

$$\begin{pmatrix}
  x' \\ y'
\end{pmatrix} = \begin{pmatrix}
  1 & b_x \\ b_y & 1
\end{pmatrix} \cdot \begin{pmatrix}
  x \\ y
\end{pmatrix}$$


```python
# Cisalhamento da imagem mario.png:
(l, c, p) = mario.shape
mario_shear = np.zeros(shape=(l+150,c+150, p), dtype=np.uint8) 
# onde shape possui l, c, p onde l é a altura, c é a largura e p é o número de canais de cor (RGB).
shear_factor = 0.5 # fator de cisalhamento
for i in range(l):
    for j in range(c):
        new_x = int(j + shear_factor * i) # nova coordenada x com cisalhamento
        new_y = i # a coordenada y permanece a mesma
        
        if new_x < c + 100: # Verifica se a nova coordenada x está dentro dos limites da imagem resultante        
            mario_shear[new_y, new_x] = mario[i, j]

plt.figure(figsize=(8, 8))
plt.imshow(mario_shear)
```

## Rotação
* Rotacionar uma imagem significa girar esta mesma imagem em α, podendo ser em graus ou radianos;
* A rotação pode ser feita a partir da origem ou de um ponto;
* Numa rotação sem interpolação pode ocorrer ausência de pixels, e em rotações em ângulos não-múltiplos de 90° pode ocorrer aliasing.

$$\begin{pmatrix}
  x' \\ y'
\end{pmatrix} = \begin{pmatrix}
  \cos\alpha & -\sin\alpha \\ \sin\alpha & \cos\alpha
\end{pmatrix} \cdot \begin{pmatrix}
  x \\ y
\end{pmatrix}$$

### Rotação

```python
alpha = np.pi / 6 # 30° de rotação
# onde ls e cs são as novas dimensões da imagem rotacionada, calculadas para garantir que a imagem caiba completamente após a rotação.
ls, cs = int(l * np.sqrt(2)), int(c * np.sqrt(2))

# Criação de uma nova imagem vazia com as dimensões calculadas para a imagem rotacionada.
mario_rot = np.zeros((ls, cs, p), dtype=np.uint8)

# Iteração sobre os pixels da imagem original,
for i in range(l):
    for j in range(c):
        x = j
        y = i
        
        # cálculo das novas coordenadas (new_x, new_y) para cada pixel usando as fórmulas de rotação.
        new_x = int(x * np.cos(alpha) - y * np.sin(alpha) +l/2) 
        new_y = int(x * np.sin(alpha) + y * np.cos(alpha))
        
        # atribuição do valor do pixel da imagem original para a nova posição na imagem rotacionada.
        mario_rot[new_y, new_x] = mario[i, j]

# figura com 8 polegadas de largura e altura, e plot da imagem rotacionada.
plt.figure(figsize=(8, 8))
# rotação sem interpolacao faltam pixels, por isso a imagem fica com buracos.
plt.imshow(mario_rot)
```

### Rotação (interpolação)

```python
alpha = np.pi / 6 # 30°

# onde ls e cs são as novas dimensões da imagem rotacionada, calculadas para garantir que a imagem caiba completamente após a rotação.
# sqrt(2) aumenta o espaço para evitar a imagem fique cortada após a rotação.
ls, cs = int(l * np.sqrt(2)), int(c * np.sqrt(2))

# Criação de uma nova imagem vazia com as dimensões calculadas para a imagem rotacionada.
mario_rot = np.zeros((ls, cs, p), dtype=np.uint8)

for i in range(ls):
    for j in range(cs):
        # cálculo das coordenadas do centro da imagem original em relação à nova imagem rotacionada.
        cx = j - (ls / 2) # coordenada x do centro da imagem original em relação à nova imagem rotacionada.
        cy = i - (cs / 2) # coordenada y do centro da imagem original em relação à nova imagem rotacionada.
        
        # cálculo das novas coordenadas (new_x, new_y) para cada pixel usando as fórmulas de rotação.  
        new_x = int( cx * np.cos(alpha) + cy * np.sin(alpha) + l / 2)
        new_y = int(-cx * np.sin(alpha) + cy * np.cos(alpha) + c / 2)
        
        # Pegar apenas a imagem do Mario na original.
        if 0 <= new_x < c and 0 <= new_y < l:
            mario_rot[i, j] = mario[new_y, new_x]

plt.figure(figsize=(8, 8))
# fica ainda um pouco mais pixelada, mas a imagem fica mais completa.
plt.imshow(mario_rot)
```

## Escala
* Escalonar uma imagem é mudar as dimensões da mesma;
* Em outras palavras, a imagem ou é "esticada" ou é "encolhida";
* A mudança de escalda ocorre pelo produto das coordenadas dos pixels pelo fator de escala.

$$\begin{pmatrix}
  x' \\ y'
\end{pmatrix} = \begin{pmatrix}
  s_x & 0 \\ 0 & s_y
\end{pmatrix} \cdot \begin{pmatrix}
  x \\ y
\end{pmatrix}$$

```python
scale_x, scale_y = 2, 2 # fatores de escala para x e y, respectivamente.
# onde l, c, p são as dimensões da imagem original. A nova imagem terá o dobro do tamanho da original em ambas as dimensões.
mario_scale = np.zeros(shape=(l * scale_y, c * scale_x, p), dtype=np.uint8)

for i in range(l):
    for j in range(c):
        new_x = np.round(j * scale_x) # nova coordenada x com escala aplicada, arredondada para o inteiro mais próximo.
        new_y = np.round(i * scale_y) # nova coordenada y com escala aplicada, arredondada para o inteiro mais próximo.
        mario_scale[new_y, new_x] = mario[i, j]

plt.imshow(mario_scale)
Image.fromarray(mario_scale).save('mario_scale.png')
```

* Para resolver o problema de ausência de pixels, faz-se uso da **interpolação**;
* Objetivo é determinar valores desconhecidos a partir dos conhecidos. O resultado estará dentro da faixa de valores conhecidos;
* Algumas técnicas incluem: nearest neighbor, interpolação linear, interpolação bilinear, interpolação bicúbica etc.

### Nearest neighbor
* Vizinho mais próximo;
* Neste método, é adotado que para cada pixel **i, j** da imagem resultante, o valor do vizinho (que fora perdido) corresponde ao do seu vizinho.


Se a imagem de entrada, $I$, é $L_{entrada}×C_{entrada}$ e a imagem de saída, $J$, é $L_{saída}×C_{saída}$, então a coordenada do pixel de saída $(l_{saída}, c_{saída})$ é amostrada da coordenada de entrada $(l, c)$ onde
$$l=\lfloor{l_f}\rfloor,$$
$$c=\lfloor{c_f}\rfloor,$$
$$l_f=l_{saída}(L_{entrada}/L_{saída}),$$
$$c_f=c_{saída}(C_{entrada}/C_{saída}).$$

```python
ls, cs = l * 2, c * 2 # nova largura e altura da imagem escalada, que é o dobro da original.
# nova imagem vazia com as dimensões calculadas para a imagem escalada.
mario_nn = np.zeros(shape=(ls, cs, p), dtype=np.uint8)

for i in range(ls):
    for j in range(cs):
        new_y = int(np.floor(i * (l / ls))) # nova coordenada y na imagem original, usando a fórmula de interpolação vizinho mais próximo.
        new_x = int(np.floor(j * (c / cs))) # nova coordenada x na imagem original, usando a fórmula de interpolação vizinho mais próximo.
        
        mario_nn[i, j] = mario[new_y, new_x]
        
plt.imshow(mario_nn)
```

## Interpolação bilinear
* Neste método, cada pixel **i, j** da imagem resultante corresponde à média ponderada dos quatro pixels vizinhos correspondentes da imagem original.

Seja $I$ a imagem de entrada $L_{entrada}×C_{entrada}$ e $J$ a imagem de saída $L_{saída}×C_{saída}.$

Seja $s_L = L_{entrada}/L_{saída}$ e $s_C = C_{entrada}/C_{saída}.$

Seja $l_f = l \cdot s_L$ para $r=0,\ldots,R_{saída}-1$ e $c_f = c \cdot s_C$ para $c=0,\ldots,C_{saída}-1.$

Seja $r_0 = \lfloor{l_f}\rfloor$ e $c=\lfloor{c_f}\rfloor.$

Seja $\Delta_l = l_f - l_0$ e $\Delta_c = c_f - c_0.$

Então $J(l, c)$
$$=I(l_0, c_0) \cdot (1-\Delta_l) \cdot (1-\Delta_c)$$
$$+I(l_0+1, c_0) \cdot \Delta_l \cdot (1-\Delta_c)$$
$$+I(l_0, c_0+1) \cdot (1-\Delta_l) \cdot \Delta_c$$
$$+I(l_0+1, c_0+1) \cdot \Delta_l \cdot \Delta_c.$$
